```python
def simulated_doctor(draft):

    prompt = f"""
You are a senior physician.

Review this discharge summary.

Rules:

1. Expand abbreviations
2. Add missing clinician review flags
3. Clarify medication instructions
4. Improve diagnosis wording

Return edited summary only.

Summary:

{draft}
"""

    response = model.generate_content(prompt)

    return response.text
```

```python
edited_report = simulated_doctor(report)

print(edited_report[:5000])
```

```python
!pip install rapidfuzz
```

```python
from rapidfuzz import fuzz

score = fuzz.ratio(
    report,
    edited_report
)

print("Similarity:", score)
```

```python
def extract_corrections(draft, edited):

    prompt = f"""
Compare:

DRAFT:
{draft}

EDITED:
{edited}

Classify corrections into:

1. Diagnosis
2. Medication
3. Follow-up
4. Safety Flags

Return bullet points.
"""

    response = model.generate_content(prompt)

    return response.text
```

```python
learning = extract_corrections(
    report,
    edited_report
)

print(learning)
```

```python
correction_memory = {
    "diagnosis": [],
    "medications": [],
    "followup": [],
    "review_flags": []
}

correction_memory["review_flags"].append(
    learning
)

print(correction_memory)
```

```python
def improved_discharge_agent(text):

    prompt = f"""
Use these learned corrections:

{learning}

Generate discharge summary.

Patient Record:

{text}
"""

    response = model.generate_content(prompt)

    return response.text
```

```python
improved_report = improved_discharge_agent(
    all_text
)
```

```python
new_score = fuzz.ratio(
    improved_report,
    edited_report
)

print("Before:", score)
print("After :", new_score)
```
